```python
df['Sale_Date'] = pd.to_datetime(df['Sale_Date'])
df.dtypes
```

```python
print("Invalid Sales Amount:", (df['Sales_Amount'] <= 0).sum())
print("Invalid Quantity Sold:", (df['Quantity_Sold'] <= 0).sum())
print("Invalid Unit Cost:", (df['Unit_Cost'] <= 0).sum())
print("Invalid Unit Price:", (df['Unit_Price'] <= 0).sum())
print("Invalid Discount:", ((df['Discount'] < 0) | (df['Discount'] > 1)).sum())
```

```python
Q1 = df['Sales_Amount'].quantile(0.25)
Q3 = df['Sales_Amount'].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
```

```python
IQR = Q3 - Q1

print("IQR:", IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[
    (df['Sales_Amount'] < lower_limit) |
    (df['Sales_Amount'] > upper_limit)
]

print("Number of Outliers:", len(outliers))
```

```python
correlation = df[
    ['Sales_Amount', 'Quantity_Sold', 'Unit_Cost', 'Unit_Price', 'Discount']
].corr()

correlation
```

```python
df.to_csv("cleaned_sales_data.csv", index=False)
```

```python
region_sales = df.groupby('Region')['Sales_Amount'].sum().sort_values(ascending=False)

region_sales
```

```python
category_sales = df.groupby('Product_Category')['Sales_Amount'].sum().sort_values(ascending=False)

category_sales
```

```python
rep_sales = df.groupby('Sales_Rep')['Sales_Amount'].sum().sort_values(ascending=False)

rep_sales
```

```python
category_quantity = df.groupby('Product_Category')['Quantity_Sold'].sum().sort_values(ascending=False)
category_quantity
```

```python
print("Total Revenue:", round(df['Sales_Amount'].sum(), 2))
print("Total Orders:", len(df))
print("Average Order Value:", round(df['Sales_Amount'].mean(), 2))

print("\nBest Region:")
print(region_sales.idxmax(), "-", round(region_sales.max(), 2))

print("\nBest Product Category:")
print(category_sales.idxmax(), "-", round(category_sales.max(), 2))

print("\nBest Sales Representative:")
print(rep_sales.idxmax(), "-", round(rep_sales.max(), 2))

print("\nHighest Quantity Category:")
print(category_quantity.idxmax(), "-", category_quantity.max())
```
